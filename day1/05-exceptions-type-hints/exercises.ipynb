## Exercise 1: Exception policy
parse_amount(raw: str) -> Decimal implement করো। Empty, invalid, negative এবং precision limit-এর জন্য meaningful exception দাও।

```python
from decimal import Decimal, InvalidOperation

class DataValidationError(ValueError):
    pass

def parse_amount(raw: str) -> Decimal:
    try:
        amount = Decimal(raw)
    except InvalidOperation as error:
        raise DataValidationError("amount must be a decimal number") from error

    if not amount.is_finite():
        raise DataValidationError("amount must be finite")
    if amount < 0:
        raise DataValidationError("amount cannot be negative")
    if amount.as_tuple().exponent < -2:
        raise DataValidationError("amount supports at most 2 decimal places")
    return amount

print(parse_amount("12.50"))
for raw in ["", "abc", "-1", "1.234", "NaN"]:
    try:
        print(parse_amount(raw))
    except DataValidationError as error:
        print(type(error).__name__, error)
```
