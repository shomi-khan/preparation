```python
# If needed: %pip install pandas
import pandas as pd

df = pd.DataFrame([
    {"transaction_id": "t1", "customer": "A", "amount": 100, "status": "success", "month": "2026-01"},
    {"transaction_id": "t2", "customer": "B", "amount": 250, "status": "failed", "month": "2026-01"},
    {"transaction_id": "t3", "customer": "A", "amount": 300, "status": "success", "month": "2026-02"},
    {"transaction_id": "t4", "customer": None, "amount": -20, "status": "success", "month": "2026-02"},
    {"transaction_id": "t1", "customer": "A", "amount": 100, "status": "success", "month": "2026-01"},
])

print(df.info())
print("duplicate IDs:", df["transaction_id"].duplicated().sum())
print("missing customers:", df["customer"].isna().sum())

clean = df.drop_duplicates(subset=["transaction_id"]).copy()
clean = clean.dropna(subset=["customer"])
clean = clean.loc[clean["amount"].ge(0)]
successful = clean.loc[clean["status"].eq("success")]

print(successful.groupby("customer")["amount"].agg(["count", "sum"]))
print(successful.groupby("month")["amount"].sum())
```
