```python
# If needed: %pip install 'pydantic>=2,<3'
from pydantic import BaseModel, Field, ValidationError, field_validator
from typing import Literal

class PredictionRequest(BaseModel):
    model_name: str = Field(min_length=1)
    features: list[float] = Field(min_length=1)
    request_id: str = Field(min_length=1)

    @field_validator("model_name", "request_id")
    @classmethod
    def strip_text(cls, value: str) -> str:
        value = value.strip()
        if not value:
            raise ValueError("value cannot be blank")
        return value

class PredictionResult(BaseModel):
    label: Literal["approve", "review", "reject"]
    confidence: float = Field(ge=0, le=1)

request = PredictionRequest(
    model_name=" baseline-v1 ", features=[0.2, 0.8], request_id=" req-1 "
)
print(request)
print(request.model_dump_json())

try:
    PredictionResult(label="unknown", confidence=1.5)
except ValidationError as error:
    print(error)
```
