## Exercise 1: Word frequency
Case-insensitive word frequency counter implement করো। Empty text, punctuation এবং top-k behavior test করো।

```python
from collections import Counter
import re

def word_frequency(text: str) -> Counter:
    words = re.findall(r"\b[\w']+\b", text.casefold())
    return Counter(words)

sample = "Python is useful. PYTHON supports data work; data matters."
counts = word_frequency(sample)
print(counts)
print(counts.most_common(2))

def positive_integers(lines):
    for line_number, line in enumerate(lines, start=1):
        value = line.strip()
        if not value:
            continue
        try:
            number = int(value)
        except ValueError as error:
            raise ValueError(f"invalid integer on line {line_number}") from error
        if number > 0:
            yield number

print(list(positive_integers(["1", "-2", "3", ""])))
```
