## Exercise 1: Concurrent I/O
asyncio.sleep দিয়ে ১০টি independent request simulate করো। Sequential ও concurrent elapsed time compare করো।

```python
import asyncio

async def fake_request(request_id: int, semaphore: asyncio.Semaphore, active: dict):
    async with semaphore:
        active["count"] += 1
        active["max"] = max(active["max"], active["count"])
        try:
            await asyncio.sleep(0.05)
            return f"response-{request_id}"
        finally:
            active["count"] -= 1

async def main():
    active = {"count": 0, "max": 0}
    semaphore = asyncio.Semaphore(3)
    results = await asyncio.gather(
        *(fake_request(i, semaphore, active) for i in range(10))
    )
    print(results)
    print("maximum concurrent requests:", active["max"])

await main()
```
